```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv(
    "sales_data_sample.csv",
    encoding="latin1"
)

# Select features
X = df[
    ["QUANTITYORDERED", "PRICEEACH", "SALES"]
]
```

```python
X = StandardScaler().fit_transform(X)

# Elbow method
errors = []

for k in range(1, 11):

    model = KMeans(
        n_clusters=k,
        n_init=10
    )

    model.fit(X)

    errors.append(model.inertia_)
```

```python
plt.plot(
    range(1, 11),
    errors,
    marker="o"
)

plt.xlabel("K")
plt.ylabel("Inertia")
plt.show()

# Choose K from elbow
model = KMeans(
    n_clusters=3,
    n_init=10
)

clusters = model.fit_predict(X)

print(clusters)
```
